# **Engagement Index Construction**

```python
# Combine:JobInvolvement,JobSatisfaction,EnvironmentSatisfaction,RelationshipSatisfaction
df['EngagementIndex'] = df['JobInvolvement'] + df['JobSatisfaction'] + df['EnvironmentSatisfaction'] + df['RelationshipSatisfaction']
display(df.head())
```

```python
#Creating  a standardized engagement score
df['EngagementScore'] = (df['EngagementIndex'] - df['EngagementIndex'].min()) / (df['EngagementIndex'].max() - df['EngagementIndex'].min())
display(df.head())
```

# **Burnout Risk Identification**

```python
#Flag employees with: OverTime = Yes ,Low WorkLifeBalance (≤2)
df['BurnoutRisk'] = ((df['OverTime'] == 'Yes') & (df['WorkLifeBalance'] <= 2)).astype(int)
display(df.head())
```

```python
#Assign burnout risk levels (Low / Medium / High)
conditions = [
    (df['BurnoutRisk'] == 1), # High Risk: OverTime = Yes and WorkLifeBalance <= 2
    ((df['OverTime'] == 'Yes') & (df['WorkLifeBalance'] > 2)), # Medium Risk: OverTime = Yes and WorkLifeBalance > 2
    ((df['OverTime'] == 'No') & (df['WorkLifeBalance'] <= 2)) # Medium Risk: No OverTime but WorkLifeBalance <= 2
]
choices = ['High', 'Medium', 'Medium']

df['BurnoutLevel'] = np.select(conditions, choices, default='Low')
display(df[['OverTime', 'WorkLifeBalance', 'BurnoutRisk', 'BurnoutLevel']].head())
```

# **Workload & Stress Analysis**


```python
# Compare engagement across Overtime vs non-overtime employees
print("Engagement across Overtime vs. Non-Overtime:")
display(df.groupby('OverTime')['EngagementScore'].mean().sort_values(ascending=False))

```

```python
# Visualize engagement across Overtime vs non-overtime employees
plt.figure(figsize=(6, 4))
sns.barplot(x='OverTime', y='EngagementScore', data=df, palette='viridis')
plt.title('Engagement Score by OverTime Status')
plt.ylabel('Average Engagement Score')
plt.xlabel('OverTime')
plt.show()

```

```python
# Compare engagement across Travel frequency categories
print("\nEngagement across Travel Frequency Categories:")
display(df.groupby('BusinessTravel')['EngagementScore'].mean().sort_values(ascending=False))

```

```python
# Visualize engagement across Travel frequency categories
plt.figure(figsize=(8, 5))
sns.barplot(x='BusinessTravel', y='EngagementScore', data=df, palette='magma')
plt.title('Engagement Score by Business Travel Frequency')
plt.ylabel('Average Engagement Score')
plt.xlabel('Business Travel')
plt.show()


```

```python
# Create 'CommuteDistanceCategory' based on median DistanceFromHome
median_distance = df['DistanceFromHome'].median()
df['CommuteDistanceCategory'] = df['DistanceFromHome'].apply(lambda x: 'Long' if x > median_distance else 'Short')

# Compare engagement across Long vs short commute distances
print("\nEngagement across Long vs. Short Commute Distances:")
display(df.groupby('CommuteDistanceCategory')['EngagementScore'].mean().sort_values(ascending=False))

```

```python
# Visualize engagement across Long vs short commute distances
plt.figure(figsize=(6, 4))
sns.barplot(x='CommuteDistanceCategory', y='EngagementScore', data=df, palette='cividis')
plt.title('Engagement Score by Commute Distance Category')
plt.ylabel('Average Engagement Score')
plt.xlabel('Commute Distance')
plt.show()
```

```python
# Define thresholds for stagnation and disengagement
# Using the 75th percentile for years to define 'long tenure/stagnation'
# Using the 25th percentile for EngagementScore to define 'low engagement'

years_in_role_threshold = df['YearsInCurrentRole'].quantile(0.75)
years_since_promotion_threshold = df['YearsSinceLastPromotion'].quantile(0.75)
low_engagement_threshold = df['EngagementScore'].quantile(0.25)

print(f"Threshold for YearsInCurrentRole (75th percentile): {years_in_role_threshold}")
print(f"Threshold for YearsSinceLastPromotion (75th percentile): {years_since_promotion_threshold}")
print(f"Threshold for Low Engagement (25th percentile): {low_engagement_threshold}")
```

```python
# Identify stagnation-linked disengagement
df['StagnationDisengagement'] = (
    (df['YearsInCurrentRole'] >= years_in_role_threshold) &
    (df['YearsSinceLastPromotion'] >= years_since_promotion_threshold) &
    (df['EngagementScore'] <= low_engagement_threshold)
).astype(int)

# Display summary statistics for this group
print("\nEmployees identified with Stagnation-Linked Disengagement:")
display(df[df['StagnationDisengagement'] == 1][['YearsInCurrentRole', 'YearsSinceLastPromotion', 'EngagementScore']].describe())
```

```python
# Count the number of employees with stagnation-linked disengagement
num_stagnated_disengaged = df['StagnationDisengagement'].sum()
total_employees = len(df)
percentage_stagnated_disengaged = (num_stagnated_disengaged / total_employees) * 100

print(f"Number of employees with stagnation-linked disengagement: {num_stagnated_disengaged}")
print(f"Percentage of employees with stagnation-linked disengagement: {percentage_stagnated_disengaged:.2f}%")

# Visualize the distribution of stagnation-linked disengagement
plt.figure(figsize=(6, 4))
sns.countplot(x='StagnationDisengagement', data=df, palette='viridis')
plt.title('Distribution of Stagnation-Linked Disengagement')
plt.xlabel('Stagnation-Linked Disengagement (1=Yes, 0=No)')
plt.ylabel('Number of Employees')
plt.xticks([0, 1], ['No', 'Yes'])
plt.show()
```
